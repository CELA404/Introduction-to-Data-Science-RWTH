# Neural Networks & Naïve Bayes

```python
import pandas as pd

import seaborn as sns
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
```

```python
df_train = pd.read_csv('train_insurance_one_hot.csv')
df_test = pd.read_csv('test_insurance_one_hot.csv')
X_train_all_feat = df_train.drop(columns=['chargeGroup', 'id','charges'], axis=1)
X_test_all_feat = df_test.drop(columns=['chargeGroup', 'id','charges'], axis=1)

y_train = df_train['chargeGroup']
y_test = df_test['chargeGroup']
```

## a)

```python
import numpy as np

classifier = MLPClassifier(hidden_layer_sizes=(5,5,5,5,5),
                           max_iter=2000, activation="logistic",random_state=42)
classifier.fit(X_train_all_feat, y_train)
Y_test_predictions = classifier.predict(X_test_all_feat)
confusion = confusion_matrix(y_test, Y_test_predictions)
accuracy = accuracy_score(y_test, Y_test_predictions)
print('Confusion Matrix of MLP:')
print(confusion)
print(f'Accuracy of MLP: {accuracy:.2f}')

```

```python
prob = classifier.predict_proba(X_test_all_feat)
class_variance = np.var(prob, axis=0) 
mask = (prob > 0.9) 
count_over_90 = np.sum(mask)

print(f'Number of instances with probabilities above 90%: {count_over_90}')
print(f'Number of total instances: {prob.shape[0]}')
print('Class_variance of MLP:', [f"{x:.2f}" for x in class_variance])
print('Probability estimates for the first five instances of the test set:')
for row in prob[:5, :]:
    print([f"{p:.2f}" for p in row])
```

## b)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_all_feat)
X_test_scaled = scaler.transform(X_test_all_feat)

classifier = MLPClassifier(hidden_layer_sizes=(5,5),
                           max_iter=2000, activation="logistic",random_state=42)
classifier.fit(X_train_scaled, y_train)
Y_test_predictions = classifier.predict(X_test_scaled)
confusion = confusion_matrix(y_test, Y_test_predictions)
accuracy = accuracy_score(y_test, Y_test_predictions)
print('Confusion Matrix of MLP:')
print(confusion)
print(f'Accuracy of MLP: {accuracy:.2f}')
```

## c)

```python
from sklearn.naive_bayes import GaussianNB


clf = GaussianNB()
# without scaling
clf.fit(X_train_all_feat,y_train)
Y_test_predictions = clf.predict(X_test_all_feat)
confusion = confusion_matrix(y_test, Y_test_predictions)
accuracy = accuracy_score(y_test, Y_test_predictions)
print('Confusion Matrix of GaussianNB without scaling:')
print(confusion)
print(f'Accuracy of GaussianNB without scaling: {accuracy:.2f}')

# with scaling
scaler = StandardScaler()
x_scaled_data_train= scaler.fit_transform(X_train_all_feat)
x_scaled_data_test= scaler.transform(X_test_all_feat)
clf.fit(x_scaled_data_train,y_train)
Y_test_predictions = clf.predict(x_scaled_data_test)
confusion = confusion_matrix(y_test, Y_test_predictions)
accuracy = accuracy_score(y_test, Y_test_predictions)
print('Confusion Matrix of GaussianNB with scaling:')
print(confusion)
print(f'Accuracy of GaussianNB with scaling: {accuracy:.2f}')
```
