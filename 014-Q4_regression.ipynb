# Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
import seaborn as sns
```

```python
train_dataset_oh = pd.read_csv('./train_insurance_one_hot.csv')
x_train = train_dataset_oh.drop(columns=['id','charges','chargeGroup'], inplace=False)
y_train = train_dataset_oh[["charges"]]

test_dataset_oh = pd.read_csv('./test_insurance_one_hot.csv')
x_test = test_dataset_oh.drop(columns=['id','charges','chargeGroup'], inplace=False)
y_test = test_dataset_oh[["charges"]]
```

## a)

```python
sns.lmplot(data=train_dataset_oh, x="age", y="charges")
plt.title("Linear regression of charges over the age")
```

## b)

```python
model = LinearRegression()
model.fit(x_train, y_train)
y_pred_train = model.predict(x_train)
mae_train = mean_absolute_error(y_train, y_pred_train)
print('Coefficients of the model')
for row in model.coef_:
    print([f"{p:.2f}" for p in row])
print(f'Mean Absolute Error of the train set:{mae_train:.2f}')

y_pred = model.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
print(f'Mean Absolute Error of the test set:{mae:.2f}')
```
